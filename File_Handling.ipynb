```python
def writing(filename,text):
  f=open(filename,'w')
  f.write(text)
  f.close()
# writing("file1.txt","hello world python")
def appending(filename,text):
  f=open(filename,'a')
  f.write(text)
  f.close()
# appending("file1.txt"," Funny language")
def reading(filename):
  try:
    f=open(filename,'r')
    text=f.read()
    print(text)
    f.close()
  except FileNotFoundError:
     print("file not found")
reading("file1.txt")
```

```python
def search(filename,word):
  try:
    f=open(filename,'r')
    line_count=0
    for line in f.readlines():
      line_count+=1
      strlist = line.split(' ')
      word_count=0
      for w in strlist:
        word_count+=1
        if word==w:
          return(line_count,word_count)
    else:
      return None
    f.close()
  except FileNotFoundError:
     print("file not found")
s="a quick brown fox jumps over the lazy dog\n"
s+="Definition: It contains all 26 letters (A through Z) in a short, easy-to-remember phrase."
writing("file1.txt",s)
t=search("file1.txt","all")
print(t)
```

```python
def modify(filename,oldwrd,newwrd):
  t=search(filename,oldwrd)
  if t!=None:
    mylist=[]
    try:
      f=open(filename,'r')
      for line in f.readlines():
        line=line.replace(oldwrd,newwrd)
        mylist.append(line)
      f.close()
      f=open(filename,'w')
      f.write(''.join(mylist))
      f.close()
    except FileNotFoundError:
      print("file not found")
  else:
    print("Search Failed")
modify("file1.txt","fox","cat")
```
