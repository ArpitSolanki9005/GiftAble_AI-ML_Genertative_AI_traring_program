```python
#  Measures of Central Tendency
import numpy as np
import pandas as pd
data = np.array([12, 15, 18, 20, 22, 25, 30, 100])  # 100 is an outlier
print("Mean:  ", np.mean(data))          # 30.25  (pulled up by outlier)
print("Median:", np.median(data))        # 21.0   (robust)
print("Mode:  ", pd.Series(data).mode()[0])
```

```python
import numpy as np
import pandas as pd
data = np.array([12, 15, 18, 20, 22, 25, 30, 100])
print("Original Data:", data)
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", pd.Series(data).mode().tolist())
```

```python
import statistics as st
rahul = [45,49,50,53,52,55]
print(st.mean(rahul))
print(st.median(rahul))
print(st.mode(rahul))
```

```python
mymarks=[76,80,79,78,75]
res = st.mean(mymarks)
print(res,end='\n')
print(st.pstdev(mymarks))
```
