```python
import numpy as np
import pandas as pd
# Correct path of the dataset
df = pd.read_csv("/content/WA_Fn-UseC_-Telco-Customer-Churn.csv")
print("Dataset successfully load ho gaya hai!")
```

```python
# get first and last 5 rows
# Pehli 5 rows dekhne ke liye
print("Pehli 5 Rows (Head)")
display(df.head())

# Aakhiri 5 rows dekhne ke liye
print("\nAakhiri 5 Rows (Tail)")
display(df.tail())
```

```python
# 3.check any missing data
print("Missing Values Per Column")
print(df.isna().sum())
```

```python
# TotalCharges ko numbers mein badlein aur khali spaces ko NaN banayein
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Ab NaN values ko us column ke Median (beech ki value) se fill karein without inplace warning
total_charges_median = df["TotalCharges"].median()
df["TotalCharges"] = df["TotalCharges"].fillna(total_charges_median)

print(f"TotalCharges filled with median: {total_charges_median}")
```

```python
# Naya feature: TotalCharges ko tenure se divide karke average nikalna
# tenure + 1 isliye kiya taaki agar tenure 0 ho toh 'Divide by Zero' ka error na aaye
df["Avg_Monthly_Spent"] = df["TotalCharges"] / (df["tenure"] + 1)

print("\nFeature Engineering Ke Baad Naya Column (Avg_Monthly_Spent)")
display(df[["customerID","tenure","TotalCharges","Avg_Monthly_Spent"]].head())
```
