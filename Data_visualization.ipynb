```python
import matplotlib.pyplot as plt
months = ['Jan','Feb','Mar','Apr','May','Jun']
sales= [10000,12000,18000,9000,23000,28000]
plt.plot(months,sales)
plt.title("line chart")
plt.xlabel("Months:")
plt.ylabel("Sales")
plt.grid()
plt.show()


```

```python
#bar chart
products = ['laptop','mobile','tv','mouse']
sales = [10000,12000,18000,9000]
plt.bar(products,sales)
plt.title("Bar chart")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
#pie chart
products = ['laptop','mobile','tv','mouse']
sales = [10000,12000,18000,9000]
plt.pie(sales,labels=products)
plt.title("Pie chart")
plt.show()
```

```python
products=["Laptop","Mobile","Monitor","Keyborad","Mouse"]
psales=[25000,35000,22000,38000,40000]
plt.pie(psales,labels=products,autopct="%1.1f%%")
plt.title("Product Sales")
plt.show()
```

### Histogram
A histogram is used to represent the distribution of a continuous dataset. It groups data into bins and shows the frequency (count) of values falling into each bin.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Generate synthetic data (e.g., heights of 250 people following a normal distribution)
np.random.seed(42)
heights = np.random.normal(loc=170, scale=10, size=250)

# Plotting the histogram
plt.figure(figsize=(8, 5))
plt.hist(heights, bins=15, edgecolor='black', color='skyblue', alpha=0.7)
plt.title("Histogram of Heights")
plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
